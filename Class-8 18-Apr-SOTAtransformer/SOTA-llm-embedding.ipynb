## Semantic Search (Similarity Search)

1. Dot Product
2. Cosine Sim
3. Eculidean Distance(L2)

```python
def dot_product(a, b):
    return np.dot(a, b)
```

```python
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
```

```python
def euclidean_distance(a, b):
    a = np.array(a)
    b = np.array(b)
    return np.linalg.norm(a - b)
```

```python
print("=== Cosine Similarity (Higher is better) ===")
print("Range: -1 to 1")

print("\n=== Dot Product (Higher is better) ===")
print("Range: -∞ to +∞ (depends on vector magnitude)")

print("\n=== Euclidean Distance (Lower is better) ===")
print("Range: 0 to +∞")
```

```python
def keyword_search(query, documents):
    query_words = query.lower().replace("?", "").split()
    results = []

    for doc in documents:
        doc_words = doc.lower().replace(".", "").split()
        doc_count = Counter(doc_words)

        score = sum(doc_count[word] for word in query_words) 
        results.append((doc, score))

    return sorted(results, key=lambda x: x[1], reverse=True)
```

```python
keyword_results = keyword_search(query, documents)
```

1. What is ResNet? (Simple Explanation)

👉 ResNet (Residual Network) is a deep learning model used for image tasks.

👉 Its main idea:

“Instead of learning full output, learn only the difference (residual).”

🧠 Core Formula (Very Important)

y=F(x)+x

Where:

x = input
F(x) = learned function
y = output

👉 This is called a Residual Block

🧩 Simple Example

Think like this:

Input = 5
Output should be = 7

Instead of learning 7, model learns:

👉 +2 (difference)

Then:
👉 Output = 5 + 2 = 7

✔ Easier learning
✔ Faster training

🔗 2. What is Skip Connection?

👉 A shortcut path that directly passes input to output

Input → Layers → + Input → Output

👉 This helps:

Avoid information loss
Train very deep networks
⚠️ 3. Problem ResNet Solves

👉 Degradation Problem

More layers → worse accuracy ❌
Training becomes difficult

👉 ResNet fixes this using:
✔ Residual learning
✔ Skip connections

🚀 4. Why ResNet is Important?
Can train very deep networks (100+ layers)
Won ImageNet 2015 competition
Used in:
Image classification
Object detection
Face recognition

Positional Encoding with example

Good question 👍 — this is one of the most confusing parts of the paper.
I’ll explain Positional Encoding in very simple English with an example.

🧠 1. Why do we need Positional Encoding?

👉 Transformer sees all words at the same time

So it does NOT know order

❌ Without position:

Sentence:

“I love you”
“You love I”

👉 For Transformer, both look same 😬

✅ Solution:

👉 Add position information to each word

This is called:

👉 Positional Encoding

📍 2. Simple idea

👉 Every word gets:

its meaning (embedding)
its position

👉 Final input =
Word meaning + Position value

🔢 3. Real Example

Sentence:
👉 “I love AI”

| Word | Position |
| ---- | -------- |
| I    | 1        |
| love | 2        |
| AI   | 3        |
Step 1: Word embeddings (simple numbers)

Let’s assume:

I → [1, 0]
love → [0, 1]
AI → [1, 1]
Step 2: Positional encoding (simple version)
Position 1 → [0.1, 0.2]
Position 2 → [0.2, 0.3]
Position 3 → [0.3, 0.4]
Step 3: Add them

👉 Final vectors:

I → [1,0] + [0.1,0.2] = [1.1, 0.2]
love → [0,1] + [0.2,0.3] = [0.2, 1.3]
AI → [1,1] + [0.3,0.4] = [1.3, 1.4]

👉 Now model knows:

word meaning
AND position
🔬 4. Actual Formula (from paper)

The paper uses sine and cosine:
PE(pos,2i)=sin(100002i/dpos​),PE(pos,2i+1)=cos(100002i/dpos​)

😌 5. Don’t worry about formula
👉 Simple meaning:


It creates unique pattern for each position


Helps model understand:


order


distance between words





🎯 6. Why sine & cosine?
👉 Because:


Smooth pattern


Model can learn relationships like:


“next word”


“2 words before”





🧠 7. Super Simple Analogy
Think of:
👉 Students sitting in class


Word = student


Position = roll number


Even if students are same,
👉 roll number tells order

🧑‍💻 One-line summary
👉 Positional Encoding = adding position information to words so Transformer understands order
