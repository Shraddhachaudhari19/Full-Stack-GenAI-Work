```python
from bs4 import BeautifulSoup

with open("page.html", "r", encoding="utf-8") as f:
    soup = BeautifulSoup(f, "html.parser")

reviews = soup.find_all("span", {"data-hook": "review-body"})

review_list = []

for r in reviews:
    # print(r)
    # print("---------------------------------------")
    review_list.append(r.text.strip())

print(review_list)
print(len(review_list))
```

```python
# STEP 1: Lowercase
review_list = [r.lower() for r in review_list]

print(review_list)
```

```python
# STEP 2: Remove punctuation
import re

def remove_punctuation(text):
    return re.sub(r'[^\w\s]', '', text)

review_list = [remove_punctuation(r) for r in review_list]

print(review_list)
```

```python
# STEP 3: Tokenization
tokens_list = [r.split() for r in review_list]

print(tokens_list)
```

```python
# STEP 4: Remove Stopwords (Optional but good)
from nltk.corpus import stopwords
import nltk

nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

tokens_list = [
    [word for word in tokens if word not in stop_words]
    for tokens in tokens_list
]

print(tokens_list)
```

TASK 2: Vocabulary Creation (VERY EASY)
🧠 What is Vocabulary?

👉 Vocabulary = all unique words from your data

🔍 Example

If your cleaned data is:

[
 ['quality', 'amazing'],
 ['good'],
 ['nice', 'dress']
]

👉 Vocabulary becomes:

['quality', 'amazing', 'good', 'nice', 'dress']

👉 No duplicates ❗


```python
tokens_list = [
    ['good', 'product'],
    ['good', 'quality'],
    ['bad', 'product']
]
```

```python
from collections import Counter

all_words = []

for tokens in tokens_list:
    all_words.extend(tokens)
    # print(tokens)
    # print(all_words)
    
print("----------------------------------------")
vocab = Counter(all_words)

print("Vocabulary:", vocab)
print("Vocabulary size:", len(vocab))
print("Top 5 words:", vocab.most_common(5))
```

Make Vocabulary

```python
# TASK 3: Feature Engineering
# Now we convert words → numbers

# 🔹 1. One Hot Encoding (Simple)

# 👉 Idea:

# If word present → 1
# If not → 0

# Convert to list
vocab_list = list(vocab.keys())
print("Vocabulary List:(Unique words)", vocab_list)

# Output:
# ['good', 'product', 'quality', 'bad']

def one_hot(doc):
    return [1 if word in doc else 0 for word in vocab_list]

ohe_vectors = [one_hot(doc) for doc in tokens_list]

print(ohe_vectors)

# Output
# [
#  [1, 1, 0, 0],   # good product
#  [1, 0, 1, 0],   # good quality
#  [0, 1, 0, 1]    # bad product
# ]
```

 TASK 4: Comparison Analysis
1. Comparison Table

| Feature        | OHE (One Hot Encoding) | BoW (Bag of Words) | TF-IDF           |
| -------------- | ---------------------- | ------------------ | ---------------- |
| Value          | 0 or 1                 | Count (0,1,2…)     | Decimal (weight) |
| Meaning        | Word present or not    | Word frequency     | Word importance  |
| Repeated words | Not counted            | Counted            | Weighted         |
| Common words   | Same importance        | Same importance    | Low importance   |
| Rare words     | Same importance        | Same importance    | High importance  |
| Use case       | Simple models          | Basic NLP          | Advanced NLP     |

✅ 🧠 Task 6: Real-world Questions
🔹 1. Why Bag of Words fails in understanding semantic meaning?

👉 Main idea:
BoW only counts words, it does NOT understand meaning ❌

🔍 Example
Sentence 1: "This product is good"
Sentence 2: "This product is excellent"

👉 Meaning:

good ≈ excellent (same meaning)

👉 But BoW treats them as:

different words ❗

So:

good ≠ excellent

👉 Model thinks they are unrelated ❌

🔍 Another example
"I love this product"
"I hate this product"

👉 BoW:

both have similar words
👉 Cannot understand opposite meaning ❌
🎯 Final Answer (write this)

“Bag of Words fails to capture semantic meaning because it treats each word independently and does not understand relationships between words. Similar words like ‘good’ and ‘excellent’ are treated as different, and it also cannot capture context or sentiment.”

🔹 2. When to use BoW and TF-IDF in industry?
✅ Use Bag of Words when:
Simple text classification
Small dataset
Fast implementation needed

👉 Examples:

Spam detection
Basic sentiment analysis
✅ Use TF-IDF when:
Need better importance of words
Want to ignore common words
Medium-level NLP tasks

👉 Examples:

Search engines
Document similarity
Review analysis

🎯 Final Answer (write this)

“Bag of Words is used in simple applications like spam detection and basic classification where speed is important. TF-IDF is used when we need better feature representation, such as in search engines, document ranking, and sentiment analysis.”

Question🔹 3. Limitations of TF-IDF in real applications
❌ 1. No semantic meaning

👉 Same problem as BoW

good ≠ excellent
❌ 2. Ignores word order
"not good" → treated similar to "good"

👉 Meaning lost ❌

❌ 3. Cannot understand context

👉 Example:

"Apple is a fruit"
"Apple is a company"

👉 TF-IDF cannot differentiate ❌

❌ 4. High dimensional data

👉 Many words → large vectors
👉 Memory + computation issue

🎯 Final Answer (write this)

“TF-IDF has limitations such as lack of semantic understanding, ignoring word order, inability to capture context, and high dimensionality. It cannot understand relationships between words and may not perform well in complex NLP tasks.”

TASK 7: Sentiment Classification

👉 Goal:

Input: review text
Output: positive (1) or negative (0)

🌱 Step 1: Create Simple Dataset

👉 Use small data first (easy to understand)

```python
review_list2 = [
    "good product",
    "excellent quality",
    "bad product",
    "very poor quality",
    "nice item",
    "waste of money"
]

labels = [1, 1, 0, 0, 1, 0]   # 1 = positive, 0 = negative
```

Step 2: Train-Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    review_list2, labels, test_size=0.3, random_state=42
)
```

Step 3: Bag of Words Mod

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression

# Convert text → numbers
cv = CountVectorizer()
X_train_bow = cv.fit_transform(X_train)
X_test_bow = cv.transform(X_test)

# Train model
model_bow = LogisticRegression()
model_bow.fit(X_train_bow, y_train)

# Accuracy
accuracy_bow = model_bow.score(X_test_bow, y_test)
print("BoW Accuracy:", accuracy_bow)
```

Step 4: TF-IDF Model

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

model_tfidf = LogisticRegression()
model_tfidf.fit(X_train_tfidf, y_train)

accuracy_tfidf = model_tfidf.score(X_test_tfidf, y_test)
print("TF-IDF Accuracy:", accuracy_tfidf)
```

📊 Step 5: Compare Results

👉 Output example:

BoW Accuracy: 0.66
TF-IDF Accuracy: 0.83
🧠 Meaning

👉 TF-IDF usually better because:

Focuses on important words
Ignores common words
🎯 Final Comparison (Write this)
Method	Accuracy	Reason
BoW	Lower	treats all words equally
TF-IDF	Higher	gives importance to useful words
🧠 Simple Explanation

👉 BoW:

Counts words
Does not know importance

👉 TF-IDF:

Finds important words
Better for prediction
🎯 Final Answer (Write in assignment)

“In this task, sentiment classification was performed using Logistic Regression. The model was trained using both Bag of Words and TF-IDF features. TF-IDF performed better than BoW because it assigns higher importance to meaningful words and reduces the weight of common words.”
